````python
from google.cloud import bigquery
````

1. Connect, .Client()

````python
client = bigquery.Client(project = "bigquery-public-data")
````

2. Dataset reference, client.dataset("dataset_name","project_name")

````python
dataset_ref = client.dataset(
    "github_repos",
    project = "bigquery-public-data"
)
````

3. Load Dataset
, get_dataset()

````python
dataset = client.get_dataset(dataset_ref)
````

4. List Tables, .list_tables(), table.table_id

````python
tables = list(client.list_tables(dataset))
# Print names of all tables in the dataset (there are four!)
for table in tables:
    print(table.table_id)
````

5. Load table, table_ref -> dataset_ref.table("table_name") ,
client.get_table(table_ref),

get_table()

````python
table_ref = dataset_ref.table("commits")
# API request - fetch the table
table = client.get_table(table_ref)
````

6. Inspect .schema, .list_rows.to_dataframe

*   List item
*   List item



````python
table.schema
client.list_rows(table, max_results=5).to_dataframe()
````

| Command                | Returns            |
| ---------------------- | ------------------ |
| `bigquery.Client()`    | Client object      |

Table.schema

| Attribute                  | Meaning                                     |
| -------------------------- | ------------------------------------------- |
| `name`                     | Column name                                 |
| `field_type`               | Data type                                   |
| `mode`                     | NULLABLE, REQUIRED, REPEATED                |
| `description`              | Human-readable description                  |
| `fields`                   | Nested sub-fields (for RECORD/STRUCT types) |
| `default_value_expression` | Default value expression (if defined)       |
